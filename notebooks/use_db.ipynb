```python
import os
import sys
sys.path.append("..")

import pandas as pd
import requests

from src.utils.paths import get_project_path
```

```python
url = "http://0.0.0.0:8001/chat"
data = {"message": "Какое влияние нанопаттерны и микропаттерны могут оказывать на пролиферацию клеток?"}

response = requests.post(url, json=data)
requests.post("http://0.0.0.0:8001/clear_history", json=response.json().get("conversation_history"))

answer = response.json().get("reply")
```

```python
print(answer)
```

```python
import re

def extract_bracket_content(text: str):
    return re.findall(r'\[(.*?)\]', text)
```

```python
df = pd.read_csv(os.path.join(get_project_path(), "data", "updated_references_links.csv"))

finde_links_in_answer = extract_bracket_content(answer)
            
new_answer = answer
for link in finde_links_in_answer:
    for i in df["filename"].values:
        if link in i:
            new_link = df[df["filename"] == i].iloc[0].link_name
            new_answer = new_answer.replace(link, new_link)
```
